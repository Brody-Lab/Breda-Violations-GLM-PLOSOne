## Code for Supporting Figure S2: stimulus pairs by animal

**Goal:** Show the stimulus pairs experienced by each publication animal in stage ≥ 4, with marker size encoding trial count.

Per-animal stimulus-pair summaries are generated by `scripts/prepare_stimulus_pair_summary.py` from the all-stage trial table (`load_all_trials_df`), restricted to the 16 modeling animals.

**Outputs:** `figures/suppS2_stimulus_pair_counts.png`.

```python
import seaborn as sns

from violations_glm.config import PSYCHOMETRIC_ANIMALS
from violations_glm.io import load_stimulus_pair_summary
from violations_glm.plotting import (
    configure_publication_dpi,
    plot_stimulus_pairs_by_animal,
    save_figure,
)

configure_publication_dpi()
sns.set_context("talk")
```

```python
summary = load_stimulus_pair_summary()
```

## Stimulus pairs by trial count

Each point is an $(S_a, S_b)$ pair for one animal in training stage ≥ 4. Marker size scales with trial count (`trial_count / 100`).

```python
fig, ax = plot_stimulus_pairs_by_animal(summary)
save_figure(fig, "suppS2_stimulus_pair_counts")
```
